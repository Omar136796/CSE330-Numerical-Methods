**Task 01**

```python
p=Polynomial([5,-4.5,0,3,0,-0.5])
print(p)


p2=p(8)
print(p2.round(6))
```

```python
p11=p.deriv(1)
p111=p11(8)

p22=p.deriv(2)
p222=p22(8)


ans=(p2-p222)/((p111)**2)
print(ans.round(6))
```

**Task 02**

```python
def X_Shapa(data_x):
    data_x=data_x[::-1]
    n=len(data_x)
    m=np.zeros((n,n))

    for r in range(n):
        n2=1
        for c in range(n):
            if n2<n:
                m[r][c]=data_x[r]**(n2)
                n2+=2
    for r in range(n):
        n3=0
        for c in range(int(n/2),n):
            if n3<n:
                m[r][c]=data_x[r]**(n3)
                n3+=2
    return m

data_x=np.array([2,3,4,5])
print(X_Shapa(data_x))
```

```python
f1=Polynomial([-4.5,0,9,0,-2.5])
x=2
h=0.1

f1_prime=f1.deriv(1)
f1_actual=f1_prime(x)

#print(f1_actual)


central_D=(f1(x+h)-f1(x-h))/(2*h)

#print(central_D)

trun_error=abs(central_D-f1_actual)

print(trun_error)
```

```python
def pluton_div_diff(x,y):
    n=len(x)
    b=np.zeros((n,n))
    for r in range(n):
        for c in range(n):
            b[r][0]=y[r]

    for col in range(1,n):
        for row in range(n-col):
            b[row][col]=(b[row+1][col-1]-b[row][col-1])/(x[row+col]-x[row])

    #b=b[0]
    return b




data_x=np.array([3,4,8])
data_y=np.array([0,15,30])
test=pluton_div_diff(data_x,data_y)
print(test)
```
